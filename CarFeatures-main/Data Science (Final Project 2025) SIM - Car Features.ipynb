### Research Question 10: How many cars are made by the brand according to the transmission type of the car  (top 10)?

```python
# Continue to explore the data to address your additional research
#   questions. Add more headers as needed if you have more questions to Answer
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Remove UNKNOWN transmission types
df_filtered = df[df["transmission_type"].str.lower() != "unknown"]

# Keep top 10 brands
top_makes = df_filtered["make"].value_counts().head(10).index
df_top = df_filtered[df_filtered["make"].isin(top_makes)]

# Crosstab for counts
pivot = pd.crosstab(df_top["make"], df_top["transmission_type"])

# Plot
plt.figure(figsize=(12,6))
pivot.plot(kind="bar", stacked=True)
plt.title("Number of Cars by Brand and Transmission Type (Excluding Unknown)")
plt.xlabel("Car Brand")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)
plt.legend(title="Transmission Type")
plt.tight_layout()
plt.show()
```

### Which fuel type provides the best fuel efficiency?

```python
# Calculate average fuel efficiency
df["avg_mpg"] = (df["city_mpg"] + df["highway_mpg"]) / 2

# Remove 'unknown' fuel category
df_filtered = df[df["engine_fuel_type"].str.lower() != "unknown"]

# Average fuel efficiency by fuel type
fuel_efficiency = (
    df_filtered
    .groupby("engine_fuel_type")["avg_mpg"]
    .mean()
    .sort_values(ascending=False)
)

# Visualization
plt.figure(figsize=(10,6))
fuel_efficiency.plot(kind="bar")
plt.xlabel("Fuel Type")
plt.ylabel("Average MPG")
plt.title("Average Fuel Efficiency by Fuel Type (Excluding Unknown)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()

```

```python
# Calculate average fuel efficiency
df["avg_mpg"] = (df["city_mpg"] + df["highway_mpg"]) / 2

# Remove 'unknown' fuel category
df_filtered = df[df["fuel_category"].str.lower() != "unknown"]

# Average fuel efficiency by fuel type
fuel_efficiency = (
    df_filtered
    .groupby("fuel_category")["avg_mpg"]
    .mean()
    .sort_values(ascending=False)
)

# Visualization
plt.figure(figsize=(10,6))
fuel_efficiency.plot(kind="bar")
plt.xlabel("Fuel Type")
plt.ylabel("Average MPG")
plt.title("Average Fuel Efficiency by Fuel Type (Excluding Unknown)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

electric vehicles provide the best fuel efficiency based on average MPG values
